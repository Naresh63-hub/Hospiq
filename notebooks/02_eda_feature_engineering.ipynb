# HospitalIQ – Phase 2: Comprehensive Exploratory Data Analysis & Feature Engineering
**Healthcare Data Analytics & Patient Segmentation Study | Hyderabad Network**

This notebook presents the clinical, demographic, operational, and financial EDA across the network, followed by non-leaking patient-level behavioral feature extraction.


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

cleaned_dir = os.path.join('data', 'cleaned')
processed_dir = os.path.join('data', 'processed')

patients = pd.read_csv(os.path.join(cleaned_dir, 'patients_cleaned.csv'), low_memory=False)
billing = pd.read_csv(os.path.join(cleaned_dir, 'billing_cleaned.csv'), low_memory=False)
admissions = pd.read_csv(os.path.join(cleaned_dir, 'admissions_cleaned.csv'), low_memory=False)
print(f'Loaded cleaned data: {len(patients):,} patients, {len(billing):,} billing transactions')
```

```python
# 1. Hospital Revenue and Average Transaction Value
hosp_rev = billing.groupby('hospital_name_cleaned').agg(
    Total_Revenue_INR=('corrected_total_amount', 'sum'),
    Avg_Transaction_INR=('corrected_total_amount', 'mean'),
    Txn_Count=('transaction_id', 'count')
).reset_index()
hosp_rev['Revenue_Cr'] = (hosp_rev['Total_Revenue_INR'] / 1e7).round(2)
print(hosp_rev)
```

```python
# 2. Patient Features Summary
features_full = pd.read_csv(os.path.join(processed_dir, 'patient_features_full.csv'), low_memory=False)
print(f'Engineered features for {len(features_full):,} active patients.')
print(features_full[['recency_days', 'visit_frequency', 'monetary_total', 'avg_length_of_stay', 'readmission_rate']].describe())
```
